```python
from bromo.evaluations.tka import eval_curve_list
from bromo.evaluations.utils import avg_fwdrev_score, prepare_dfs
```

### 1. TKA plots for: yeast-exploris

```python
bromo_preds_path_exploris = '../../Datasets/inference/yeast-exploris/full_predictions.tsv'
xgboost_preds_path_exploris = '../../Datasets/inference/yeast-exploris/xgboost_predictions.tsv'
bromo_preds_exploris, xgboost_preds_exploris = prepare_dfs(bromo_preds_path_exploris, xgboost_preds_path_exploris)
bromo_uniquepair_exploris = avg_fwdrev_score(bromo_preds_exploris)
xgboost_uniquepair_exploris = avg_fwdrev_score(xgboost_preds_exploris)
```

```python
curve_dicts_q1_exploris_errorbars = eval_curve_list(
    dfs = [bromo_uniquepair_exploris,xgboost_uniquepair_exploris],
    topk=5, metric="q1", labels=['Bromo','XGBoost'],
    title = 'yeast-exploris',
    save_path = './fig2b_exploris_errorbars_sem.pdf',
    ttest_loc = (0.5, 0.45),
    legend_loc = (0.01, 0.72),
    ylim = (0,0.7),
    ttest_text_fs = 10,
    legend_fs = 12,
    show_chance = True,
    error_bars = True,
    band='sem',
    markersize=1,    # default 4
    linewidth=1.5)   # default 2.5)
```

```python
del bromo_uniquepair_exploris, xgboost_uniquepair_exploris
```
