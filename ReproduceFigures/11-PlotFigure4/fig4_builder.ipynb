```python
from bromo.evaluations.tka import eval_curve_list
from bromo.evaluations.utils import prepare_dfs, avg_fwdrev_score
```

### 1. TKA plots for: yeast-lumos

```python
scratch_preds_path_lumos = '../../Datasets/finetuning/yeast-lumos/trainedfromscratch/testset_predictions.tsv'
finetuned_preds_path_lumos = '../../Datasets/finetuning/yeast-lumos/finetuned/testset_predictions.tsv'
pretrained_preds_path_lumos = '../../Datasets/finetuning/yeast-lumos/pretrained/testset_predictions.tsv'
scratch_preds_lumos = prepare_dfs(scratch_preds_path_lumos)
finetuned_preds_lumos = prepare_dfs(finetuned_preds_path_lumos)
pretrained_preds_lumos = prepare_dfs(pretrained_preds_path_lumos)


scratch_uniquepair_lumos = avg_fwdrev_score(scratch_preds_lumos)
finetuned_uniquepair_lumos = avg_fwdrev_score(finetuned_preds_lumos)
pretrained_uniquepair_lumos = avg_fwdrev_score(pretrained_preds_lumos)
```

```python
curve_dicts_q1_lumos = eval_curve_list(
    [finetuned_uniquepair_lumos,scratch_uniquepair_lumos,pretrained_uniquepair_lumos], 
    topk=5, metric="q1", labels=['Finetuned','TrainedFromScratch','Pretrained'],
    title = 'yeast-lumos',
    save_path = './Fig4e_lumos_yeast.pdf',
    ttest_loc = (0.35, 0.37),
    ylim = (0.25, 0.7),
    legend_loc = (0.01, 0.72),
    figsize = (4,4),
    legend_fs = 12,
    colors = ["#009E73","#CC79A7","#0072B2"])
```
